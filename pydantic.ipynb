### Pydantic Basics : Creating and Using Models

```python
from pydantic import BaseModel
```

```python
## Define Person class and inherit BaseModel in it
class Person(BaseModel):
    name: str
    age: int
    city: str

person = Person(name="Anuraj", age = 23,city='kathmandu')
print(person)
```

```python
person1 = Person(name='anuraj', age = 23, city=25)
print(person1)
```

### Model with Optional Fields
 Add optional fields using Python's Optional type:

```python
from typing import Optional
class Bus(BaseModel):
    id: int
    name: str
    route: str
    AC: Optional[bool] = True  # Optinal with default value
    seat: Optional[str] = 'Sofa' # Optinal with default value
```

```python
# Example with optional fieldj
bus1 = Bus(id=2022, name='rudraksha',route='kathmandu-nepalgunj')
print(bus1)
```

```python
# Example with list. Datatype can be list also
class Hospital(BaseModel):
    id:int
    ward_no:str
    patients:list[str]
    capacity: int
```

```python
data1 = Hospital(
    id= 2555,
    ward_no = 'Ortho_101',
    patients = ['ram','shyam','hari'],
    capacity = 25
)
print(data1)
```

#### Model with Nested Models
Create complex structues with nested models:

```python
class Address(BaseModel):
    district: str
    city: str
    ward_no: int

class Patients(BaseModel):
    id: int
    name: str
    address: Address #nested model

patient1 = Patients(
    id =1111,
    name= 'ram bahadur chaurasiya',
    address={
        'district': 'bardiya' , 
        'city':'madhuban',
        'ward_no': 6
    }
)
print(patient1)
```

#### Pydantic Fields: Customizationa and Constraints
The Field function in Pydantic enhances model fields beyond basic type hinds by allowing you to specify validation rules, default values, aliases, and more. Here's a comprehensive tutorial with examples.

```python
from pydantic import BaseModel, Field
```

```python
class Item(BaseModel):
    name:str=Field(min_length=2, max_length=50) # we define constraints on top of type validation. i.e for name it should be string between length 2 and 50
    price: float = Field(gt=0, le=1000) # price should be in between 0 and 1000 of float type
    quantity: int = Field(gt=0) # quantity should be greater than 0

item = Item(name='tshirt', price='550', quantity=5)
print(item)
```
