```python
# CLIP Model
import os
from dotenv import load_dotenv
load_dotenv()
from google import genai

hf_token= os.getenv("HF_TOKEN")


# setup the environment
# os.environ["OPENAI_API_KEY"] = os.getenv("OPEN_API_KEY")

gemini_api = os.getenv("GEMINI_API_KEY")
# os.environ["GOOGLE_API_KEY"] = gemini_api
# os.environ["GOOGLE_API_KEY"] = os.getenv("GEMINI_API_KEY")
# iNITIalize the clop model for unified embeddings
clip_model=CLIPModel.from_pretrained("openai/clip-vit-base-patch32")
clip_processor=CLIPProcessor.from_pretrained("openai/clip-vit-base-patch32")

# what is this processor ? --> for giving ip to any of the model this processor makes sure that whatever format is required it will convert to that format.
# for clip model we need to import clipmodel and processor
clip_model.eval() 
```

```python
# step 3
# how to embed the images. take image part and convert into image vector
# we have to create two function. 1. def embed(image_data) 2.
def embed_image(image_data):
    """Embed image using CLIP"""
    if isinstance(image_data, str): #if image path is passed then open image an convert to rgb and store in image
        image = Image.open(image_data).convert("RGB")
    else: # If PIL Image, directly image is passed then store in image
        image = image_data

    #to convert image into image vector we use clip processor. and return tensors in pytorch tensors.
    #it converts entire format into tensors
    inputs = clip_processor(images=image, return_tensors='pt')
    with torch.no_grad():
        features = clip_model.get_image_features(**inputs)

        # In case it returns BaseModelOutputWithPooling (object) instead of tensor. we should extract .pooler_output
        if not isinstance(features, torch.Tensor):
            features = features.pooler_output

        # normalize embeddings to unit vector
        features = features / features.norm(p=2,dim=-1, keepdim=True)
        return features.squeeze().cpu().detach().numpy()
    

def embed_text(text):
    """Embed text using CLIP"""
    inputs = clip_processor(
        text=[text],
        return_tensors="pt",
        padding=True,
        truncation=True,
        max_length=77  #clip's max token length
    )

    with torch.no_grad():
        features = clip_model.get_text_features(**inputs)

        # In case it returns BaseModelOutputWithPooling
        if not isinstance(features, torch.Tensor):
            features = features.pooler_output

        #normalize embeddings
        features = features / features.norm(p=2,dim=-1, keepdim=True)
        return features.squeeze().cpu().detach().numpy()
        
```

```python
import io
from PIL import Image
# from langchain.schema import Document

## process pdf
pdf_path = "sample.pdf"
doc=fitz.open(pdf_path)

# create a variable which will store all documents and embeddings
all_docs = []
all_embeddings = []
image_data_store = {} # store actual image data for LLM

# text splitter 
splitter = RecursiveCharacterTextSplitter(chunk_size=500, chunk_overlap=100)

# iterate through all documents. get text data --> convert in text chunks --> create embeddings
for i,page in enumerate(doc):
    ## process text
    text = page.get_text()
    if text.strip():
        ## create temporary document for splitting. and keep in form of document data structure
        temp_docs = Document(page_content=text, metadata={"page":i, "type":"text"}) #for all text data keep metadata type=text
        text_chunks = splitter.split_documents([temp_docs])

        # embed each chunk using CLIP
        for chunk in text_chunks:
            embedding = embed_text(chunk.page_content)
            all_embeddings.append(embedding)
            all_docs.append(chunk)

## process images
# --three important actions:
# --convert pdf image to PIL format
# --store as base64 for gemini 
# --create clip embedding for retrieval

for i,page in enumerate(doc):
    ## process text
    text=page.get_text()
    if text.strip():
        ##create temporary document for splitting
        temp_doc = Document(page_content=text, metadata={"page": i, "type": "text"})
        text_chunks = splitter.split_documents([temp_doc])

        #Embed each chunk using CLIP
        for chunk in text_chunks:
            embedding = embed_text(chunk.page_content)
            all_embeddings.append(embedding)
            all_docs.append(chunk)



    ## process images
    ##Three Important Actions:

    ##Convert PDF image to PIL format
    ##Store as base64 for GPT-4V (which needs base64 images)
    ##Create CLIP embedding for retrieval

    for img_index, img in enumerate(page.get_images(full=True)):
        try:
            xref = img[0]
            base_image = doc.extract_image(xref)
            image_bytes = base_image["image"]
            
            # Convert to PIL Image
            pil_image = Image.open(io.BytesIO(image_bytes)).convert("RGB")
            
            # Create unique identifier
            image_id = f"page_{i}_img_{img_index}"

            #gemini will handle bytes directly            
            image_data_store[image_id] = pil_image
            
            # Embed image using CLIP
            embedding = embed_image(pil_image)
            all_embeddings.append(embedding)
            
            # Create document for image
            image_doc = Document(
                page_content=f"[Image: {image_id}]",
                metadata={"page": i, "type": "image", "image_id": image_id}
            )
            all_docs.append(image_doc)
            
        except Exception as e:
            print(f"Error processing image {img_index} on page {i}: {e}")
            continue

doc.close()
```

```python
from langchain_google_genai import ChatGoogleGenerativeAI
import os
from dotenv import load_dotenv
load_dotenv()
os.environ["GOOGLE_API_KEY"] = os.getenv('GEMINI_API_KEY')
# initialize gemini 2.5 flash
llm = llm = ChatGoogleGenerativeAI(model="gemini-2.5-flash")
llm
```

```python
# how search is happens or retrieval function/model
def retrieve_multimodal(query, k=5):
    """unified retrieval using CLIP embeddings for both text and images."""
    # Embed query using CLIP
    query_embedding = embed_text(query) # we embed the text(query) with query embedding

    # search based on the query embeddings in unified vector store
    results = vector_store.similarity_search_by_vector(
        embedding=query_embedding,
        k=k
    )
    return results
```

```python
# now convert for the function
from langchain_core.messages import HumanMessage

def create_multimodal_message(query, retrieved_docs):
    """Create a multimodal message for Gemini 2.5 Flash"""

    content = []

    # Add question
    content.append({
        "type": "text",
        "text": f"Question: {query}\n\nContext:\n"
    })

    # Separate text and image docs
    text_docs = [doc for doc in retrieved_docs if doc.metadata.get("type") == "text"]
    image_docs = [doc for doc in retrieved_docs if doc.metadata.get("type") == "image"]

    # Add text context
    if text_docs:
        text_context = "\n\n".join([
            f"[Page {doc.metadata['page']}]: {doc.page_content}"
            for doc in text_docs
        ])
        content.append({
            "type": "text",
            "text": f"\nText excerpts:\n{text_context}\n"
        })

    # Add images (DIRECT PIL IMAGES)
    for doc in image_docs:
        image_id = doc.metadata.get("image_id")

        if image_id and image_id in image_data_store:
            content.append({
                "type": "text",
                "text": f"\n[Image from page {doc.metadata['page']}]:"
            })

            content.append({
                "type": "image",
                "image": image_data_store[image_id]  # PIL Image directly
            })

    # Final instruction
    content.append({
        "type": "text",
        "text": "\nPlease answer the question based on the provided text and images."
    })

    return HumanMessage(content=content)
```

```python
def multimodal_pdf_rag_pipeline(query):
    """Main pipeline for multimodal RAG."""
    # Retrieve relevant documents
    context_docs = retrieve_multimodal(query, k=5)
    
    # Create multimodal message
    message = create_multimodal_message(query, context_docs)
    
    # Get response from gemini
    response = llm.invoke([message])
    
    # Print retrieved context info
    print(f"\nRetrieved {len(context_docs)} documents:")
    for doc in context_docs:
        doc_type = doc.metadata.get("type", "unknown")
        page = doc.metadata.get("page", "?")
        if doc_type == "text":
            preview = doc.page_content[:100] + "..." if len(doc.page_content) > 100 else doc.page_content
            print(f"  - Text from page {page}: {preview}")
        else:
            print(f"  - Image from page {page}")
    print("\n")
    
    return response.content
```

```python
if __name__ == "__main__":
    # Example queries
    queries = [
        "What does the first chart of document tells about?",
        "Summarize the main findings from the document",
        "What visual elements are present in the document?"
    ]
    
    for query in queries:
        print(f"\nQuery: {query}")
        print("-" * 50)
        answer = multimodal_pdf_rag_pipeline(query)
        print(f"Answer: {answer}")
        print("=" * 70)
```
